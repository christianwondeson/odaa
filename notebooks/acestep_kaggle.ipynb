Generates 3 instrumental score candidates (`45–60s`, slow tempo in metadata) for auditioning, then exports `chosen.wav`.

```python
import pathlib, shutil, time
from acestep.inference import generate_music, GenerationParams

out_dir = pathlib.Path("/kaggle/working/music_candidates"); out_dir.mkdir(parents=True, exist_ok=True)
caption = "warm acoustic strings, soft plucked Ethiopian krar-like lyre, sparse, gentle, intimate, no strong lead melody, instrumental"

for idx, seed in enumerate((101, 102, 103), 1):
    t0 = time.time()
    params = GenerationParams(
        prompt=caption,
        lyrics="[instrumental]",
        task_type="text2music",
        bpm=72,
        num_inference_steps=8,
        duration=60.0,
        seed=seed,
    )
    res = generate_music(params)
    dst = out_dir / f"candidate_{idx}.wav"
    shutil.copy(res["path"], dst)
    print(f"candidate_{idx}.wav generated in {time.time()-t0:.1f}s (seed={seed})")
```
